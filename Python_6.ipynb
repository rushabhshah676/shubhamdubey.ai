```python
import matplotlib.pyplot as plt
import pandas as pd

# Load the Excel file using pd.read_excel
df = pd.read_excel('/content/sample_data/Sample - Superstore.xlsx')

# Aggregate sales by Category to create a valid bar chart
category_sales = df.groupby('Category')['Sales'].sum().reset_index()

plt.bar(category_sales['Category'], category_sales['Sales'])
plt.title('Superstore Category Sales')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.show()
```
